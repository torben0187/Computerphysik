<td style="border-style: none; width: 1%; text-align: left; font-size: 25px; font-weight: bold; text-decoration: underline">LR-Zerlegung (d.h., LU-Zerlegung) </td>


In dieser Aufgabe soll für die Lösung von linearen Gleichungssystemen die sogenannte **LR**-Zerlegung implementiert werden, die Sie in der Vorlesung als **LU**-Zerlegung kennengelernt haben (**L**inks-**R**echts, oder **L**ower-**U**pper).
Die LR-Zerlegung ist im Prinzip das Gaußsche Eliminationsverfahren, das Sie wahrscheinlich schon aus der Schule kennen, nur dass die Umformungsschritte in der Matrix $L$ gespeichert werden. Das Ergebnis einer LR-Zerlegung einer Matrix $A$ ist eine untere Dreiecksmatrix mit einsen auf der Diagonalen $L$ und eine obere Dreiecksmatrix $R$ für die gilt $A = L \cdot R$ (bzw. $A = L \cdot U$ in der Notation der Vorlesung).

Solch eine Zerlegung ist sinnvoll für die Lösung von Gleichungssystemen. Betrachten wir dazu ein Gleichungsystem $Ax=b$ mit LR-Zerlegung $A = LR$. Dies kann man schreiben als

$$ L \underbrace{Rx}_{y} = b .$$

Somit kann eine Lösung des Gleichungssystems durch die folgenden zwei Schritte erfolgen:

1. Bestimme die Lösung $y$ der Gleichungssystems $Ly = b$ mit $y = Rx$ durch Vorwärtseinsetzen

2. Bestimme die Lösung $x$ des Gleichungssystems $Rx=y$ durch Rückwertseinsetzen.


Da $L$ und $R$ jeweils untere bzw. obere Dreiecksmatrizen sind, können die Gleichungssysteme sehr effizient für beliebige $b$ durch Vorwärts- bzw. Rückwertseinsetzen gelöst werden. 

Das Vorgehen bei einer LR-Zerlegung haben Sie im Detail in dem Vorlesungsvideo 4-c mithilfe eines konkreten Beispiels gesehen: Falls noch nicht gemacht, schauen Sie es bitte da nochmal an.

Wir starten dafür mit der Matrix $R_0 = A$ und anschließend gehen wir iterativ vor.
Im $i$-ten Schritt müssen nun die Zahlen der $i$-ten Spalte unter der Diagonale auf 0 gesetzt werden. Dies geschieht durch die Addition von Vielfachen der $i-$ten Zeile auf die unteren Zeilen. 
Genau diese Operation kann als Multiplikation mit einer linken unteren Dreiecksmatrix $(L_i)^{-1}$ geschrieben werden:
$$
L_i^{-1} = \begin{pmatrix}
1 & & & & & \\
& \ddots & & & & \\
& & 1 & & & \\
& & -l_{i+1, i} & 1 & & \\
& & \vdots & & \ddots & \\
& & -l_{N, i} & & & 1
\end{pmatrix} 
\Longleftrightarrow
L_i = \begin{pmatrix}
1 & & & & & \\
& \ddots & & & & \\
& & 1 & & & \\
& & l_{i+1, i} & 1 & & \\
& & \vdots & & \ddots & \\
& & l_{N, i} & & & 1
\end{pmatrix}
$$
Die Elemente $l_{ji}$ der Matrix $L_i$ sind dabei im i-ten Schritt gegeben durch 
$$l_{ji} = \frac{(R_{i-1})_{ji}}{(R_{i-1})_{ii}} \quad \forall j>i.$$

Damit lässt sich die ursprüngliche Matrix als 
$$A = R_0 = L_1 R_1 = L_1 (L_2 R_2) = \left(\prod_{i=1}^{N-1} L_i\right) R_{N-1} \equiv L R$$
umschreiben. 
Besonders einfach ist dies, da das Produkt für untere Dreiecksmatrizen einfach der Summe entspricht, ausgenommen der Diagonale:
$$
\prod_{i=1}^{N-1} L_i = 
\begin{pmatrix}
1 & & & & & \\
l_{2, 1}& \ddots & & & & \\
&\ddots & 1 & & & \\
\vdots & & l_{i+1, i} & 1 & & \\
& & \vdots & \ddots & \ddots & \\
l_{N, 1}& \ldots & l_{N, i} & \ldots & l_{N, N-1} & 1
\end{pmatrix}
$$

### b) Implementierung der LR-Zerlegung

Ergänzen sie die Funktion `LR_Zerlegung`, sodass sie die LR-Zerlegung einer beliebigen quadratischen Matrix `A` berechnet. 

*Anmerkung*: Sie sollen davon ausgehen, dass keine Zeilenvertauschungen im Algorithmus nötig sind.  
*Anmerkung*: Sollten sie hier keine funktionierende Lösung finden, können sie alle nachfolgenden Aufgaben auch mit der Julia-Funktion  
`lu(A, val{false})` lösen.  Das Argument `val{false}` ist dafür da, dass keine Zeilen während der LR-Zerlegung vertauscht werden. Das ist normalerweise nicht sinnvoll, doch hier für den Vergleich schon.

```julia
using LinearAlgebra 

function LR_Zerlegung(A)
    
    #Anzahl an Zeilen/Spalten
    N = size(A, 1)
    
    #L Matrix als Einheitsmatrix initialisiert
    L = Matrix{Float64}(I, N, N)
    
    #R Matrix wird initialisiert als A
    R = copy(A)

    #Schleife über die Zeilen von A (Letzte Zeile nicht nötig)

    #j-te Spalte
        for j in 1:N-1

    #L Wert der i-te Zeile
            for i in j+1:N
                L[i,j]=R[i,j]/R[j,j]

    #Zeilen addieren
                for k in 1:N
                    R[i,k]=R[i,k]-L[i,j]*R[j,k]
                end
            end
        end
    return L, R
end
```

Testen sie nun ihre Funktion durch die nachfolgenden Zellen.

```julia
L, R = LR_Zerlegung(A)
display(L)
display(R)
display(L*R ≈ A)
```

### c) Vorwärts und Rückwerts einsetzen

Um nun ein Gleichungssystem der Form $Ax = b$ mittels LR-Zerlegung zu lösen, muss, wie oben schon beschrieben, noch das Vorwärts bzw. Rückwertseinsetzen implementiert werden. 

Mit Vorwärtseinsetzen ist gemeint, die Lösung $y$ eines Gleichungssystem der Form

$
\begin{pmatrix}
l_{11} & 0 & 0  & \ddots & 0 \\
l_{21} & l_{22} & 0 & & \vdots \\
l_{31} & l_{32} & l_{33} & \ddots & \vdots \\
\vdots & \vdots & \vdots & \ddots & 0 \\
l_{N1} & l_{N2} & l_{N3} & \ddots & l_{NN}
\end{pmatrix}
\cdot
\begin{pmatrix} y_1 \\ y_2 \\ y_3 \\ \vdots \\y_n \end{pmatrix} =
\begin{pmatrix} b_1 \\ b_2 \\ b_3 \\ \vdots \\b_n \end{pmatrix}
$

zu berechnen, in dem man zunächst aus der ersten Zeile 

$y_1 = \frac{b_1}{l_{11}}$ 

bestimmt, das Ergebnis in die Zweite Zeile einsetzt und 

$y_2 = \frac{b_2 - y_1 l_{21}}{l_{22}}$ 

bestimmt, usw ..

Die Formel für ein allgemeines $i$ lautet:

$ y_i = \frac{1}{l_{ii}}\left(b_i - \sum_{k=1}^{i-1} l_{ik} y_k \right). $
