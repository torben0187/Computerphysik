Um die Geschwindigkeit $v$ eines Autos zu messen, wird mittels eines Laser-Entfernungsmessgeräts in Abständen von einer Sekunde jeweils der Abstand des Autos zum Beobachter $s$ gemessen. Aus diesen Abstandmessungen soll nun mittels der Gleichung 

$s(t) = v \cdot t + s_0$

und **linearer Regression** die Geschwindigkeit des Autos bestimmt werden. Wir nehmen dabei an, dass das Auto eine konstante Geschwindigkeit hat, und dass die Messung des Abstands genau in Fahrtrichtung des Autos stattfindet.
Die Abstandmesswerte sind nachfolgend im Vektor `s` und die Zeitpunkte im Vektor `t` gespeichert.

```julia
#Messzeitpunkt in Sekunden
t = collect(1:20)

#Abstandsmesswerte in Metern
s = [20.21, 44.31, 61.54, 59.5, 84.18, 95.83, 98.65, 126.82, 127.07, 149.65, 169.91, 166.8, 187.15, 194.81, 214.78, 241.59, 245.7, 253.74, 275.61, 295.41];
```

Es soll nun an die $N=20$ Messwertpaare $(t_i, s_i)$ eine Gerade der Form 

$$y(t; a_1, a_2) = a_1 + a_2 \cdot t$$

angelegt werden, die die Messwerte bestmöglich beschreibt. Der so bestimmte Paramter $a_2$ ergäbe dann die Geschwindigkeit des Autos. Da die Messwerte gewisse Messungenauigkeiten aufweisen, wird die Gerade in der Regel nicht alle Messpunkte exakt treffen. Stattdessen ist das Ziel der linearen Regression die Paramter $a_1$ und $a_2$ so zu bestimmen, dass die Summe der Fehlerquadrate

$$\min_{a_1, a_2}\sum_{i=1}^N (y(t_i; a_1, a_2) - s_i)^2 $$

minimiert wird. Dieses Minimierungsproblem kann auf unterschiedliche Arten gelöst werden, welche nun in dieser Aufgabe behandelt werden sollen.

### b) Lineare Regression als überbestimmtes Gleichungssystem

Die Aufgabe der linearen Regression kann auch als überbestimmtes Gleichungssystem formuliert werden. Für exakte Übereinstimmung der Regressionsgeraden mit den $N$ Messwertpaaren müssten folgende Gleichungen gelten:

$$ y(t_1; a_1, a_2) = a_1 + a_2 \cdot t_1  = s_1$$
$$ y(t_2; a_1, a_2) = a_1 + a_2 \cdot t_2  = s_2$$
$$ \vdots $$
$$ y(t_N; a_1, a_2) = a_1 + a_2 \cdot t_N  = s_N$$

Da hier nur zwei unbekannte $a_1$ und $a_2$, aber $N$ Gleichungen vorhanden sind, hat solch ein Gleichungssystem in der Regel keine exakte Lösung. Auch hier kann als beste Lösung die der minimalen Fehlerquadrate gewählt und mit mitteln der linearen Alebgra ermittelt werden.

Hier benötigen wir nur die Eigenschaften solch einer Zerlegung.
