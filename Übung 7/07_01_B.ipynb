## Aufgabe 1 Teil B) Lösung mit linearer Algebra

---

In dieser Aufgabe soll das Problem aus der `Aufgabe 07_01_A` mithilfe von Diagonalisierung einer Matrix gelöst werden.
Dafür benutzen Sie die Approximation der zweifachen Ableitung  nach $x$:

$$
    \left(\partial^2_x f\right)|_x \rightarrow \frac{f(x+\epsilon) - 2f(x) + f(x-\epsilon)}{\epsilon^2}
$$

Mithilfe dieser Ersetzung können Sie nun die SSG mit den Randbedingungen $\psi(x\rightarrow \pm \inf) = 0$ 

$$
    H\psi(x) = E \psi(x)
$$

in eine einfache Matrix-Eigenwertgleichung umformen

$$
    \mathcal{H}\vec{\psi} = E \vec{\psi}\,. \qquad (1)
$$


Dafür wird wieder die reelle Achse $\mathbb{R}$ mithilfe von $N$ Punkten diskretisiert:
$$
    x \in \mathbb{R} \rightarrow 
        \begin{cases} 
            x_j & := x_{min} + (j-1)\epsilon \\
            x_1 & := x_{min} \\
            x_N & := x_{min} + (N-1)\epsilon := x_{max}
        \end{cases}
$$
wobei $x_{min}$ und $x_{max}$ die Punkte im unendlichen repräsentieren mit $\psi(x_{min}) = \psi(x_{max}) = 0$

Auf dieser Diskretisierung wird dann der Lösungsvektor $\vec{\psi}$ definiert:

$$
    \vec{\psi} = (\psi_1, \psi_2, \dots, \psi_N)^T\,, \quad \psi_j := \psi(x_j)
$$

Als Letztes muss nun nur noch die obige Diskretisierung der zweifachen Ableitung in die SSG eingesetzt werden.

Als finales Ergebnis erhält man dann für die Matrix $\mathcal{H}$:

$$
\mathcal{H} = \begin{pmatrix}
                    2c + V_1 & -c       & 0        & 0  & 0            & \cdots       & 0          \\
                    -c       & 2c + V_2 & -c       & 0  & 0            & \cdots       & 0          \\
                    0        & -c       & 2c + V_3 & -c & 0            & \cdots       & 0          \\
                    \vdots   &          & \ddots   &    &\ddots        &              & \vdots     \\
                    0        & \cdots   & 0        & -c & 2c + V_{N-2} & -c           & 0          \\
                    0        & \cdots   & 0        & 0  & -c           & 2c + V_{N-1} & -c         \\
                    0        & \cdots   & 0        & 0  & 0            & -c           & 2c + V_{N}
              \end{pmatrix}
$$
mit $c = \frac{1}{2m\epsilon^2}$ und $V_j := V(x_j)$.
