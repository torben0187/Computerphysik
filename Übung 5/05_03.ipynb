<td style="border-style: none; width: 1%; text-align: left; font-size: 25px; font-weight: bold; text-decoration: underline">Die kranke Hilbertmatrix</td>


Eine Hilbertmatrix ist eine Matrix der Form $H_{ij} = 1/(i + j -1)$. Zum Beispiel ist die eine $4\times 4$ Hilbertmatrix gegeben durch

$$
\begin{pmatrix}
1 & 1/2 & 1/3 & 1/4 \\
1/2 & 1/3 & 1/4 & 1/5 \\
1/3 & 1/4 & 1/5 & 1/6 \\
1/4 & 1/5 & 1/6 & 1/7
\end{pmatrix}.
$$

Diese Art von Matrix wird häufig zum testen numerischer Algorithmen in der Linearen Algebra genutzt, da es bekannt ist, dass sie sehr *merkwürdiges* Verhalten aufweist.

### a) Hilbertmatrix generieren

Schreiben Sie nun eine Funktion, die eine $N \times N$ Hilbertmatrix generiert. Testen Sie ihren Code für verschiedene Werte von N.

```julia
function Hilbert(n)
    H = zeros(Float64, n, n)
    for i in 1:n
        for j in 1:n
            H[i,j] = 1/(i+j-1)
        end
    end
    return H
end 
```

```julia
N = 50
H = Hilbert(N)
```

### b) Gleichungssystem lösen
Lösen Sie nun das Gleichungssystem $Hx = b$ für N = 50, wobei $b$ ein Vektor aus Einsen ist. Nutzen Sie dazu die Methode ihrer Wahl. Speichern Sie das Ergebnis in einer Variable `x`.

```julia
b = ones(Float64, N)
x = H \ b
```

### c) Kleine Änderung, große Auswirkung

Erstellen Sie nun ein neues Gleichungssystem, indem Sie den ersten Eintrag von `b` um nur `1e-15` verändern. Lösen Sie das neue Gleichungssystem und speichern sie die Lösung im Vektor `x_new`. Berechnen Sie anschließend die maximale Abweichung zwischen `x_new` und `x` mittels des unten vorgegebenen Codes. Was fällt Ihnen auf?

```julia
using LinearAlgebra

b[1] = 1 + 1e-15

x_new = H \ b

#Maximale Abweichung berechnen:
abweichung = maximum(abs.(x_new - x))
println("Die maximale Abweichung beträgt: ", abweichung)
```

Erweitern Sie nun den nachfolgenden Code, um obige Rechnung für verschiedene $N$ auszuführen und anschließend die maximale Abweichung für jedes $N$ im Vektor `abweichungen` zu speichern.

```julia
N_Werte = 5:10:100

#Array zum speichern der maximalen Abweichung
abweichungen = zeros(length(N_Werte))

#Iteration über alle N
for i in 1:length(abweichungen)
    
    #Aktuelles N festlegen
    N = N_Werte[i]
    
    H = Hilbert(N)
    b = ones(N)
    x = H \ b
    b[1] = 1+1e-15
    x_new = H \ b
    abweichungen[i] = maximum(abs.(x_new - x))
end
```

Erstellen Sie nun einen einfachen Plot, der die Abweichungen gegen die N-Werte aufträgt.

```julia
using Plots

plot(N_Werte , abweichungen)
```

### d) Die Inverse
Wir wissen das $H H^{-1} = \mathbb{1}$ gelten sollte. Doch auch hier verhält sich die Hilbertmatrix bei numerischen Rechnungen sehr ungezogen. Sie sollen nun für verschiedene $N$ wieder $H$ generieren, dann die Inverse $H^{-1}$ mittels `inv(H)` bilden und anschließend $H\cdot H^{-1}$ berechnen. Berechnen Sie daraufhin für jedes $N$ die Norm der Differenz zur Einheitsmatrix $||  H H^{-1} - \mathbb{1} ||$  und speichern Sie das Ergebnisse in dem vordefinierten `abweichungen` Array.

*Hinweis:* Die Norm der Differenz einer Matrix `A` zur Einheitsmatrix kann mittels `norm(A - I)` berechnet werden.

```julia
N_Werte = 5:1:20

#Array zum speichern der maximalen Abweichung
abweichungen = zeros(length(N_Werte))

#Iteration über alle N
for i in 1:length(abweichungen)
    
    #Aktuelles N festlegen
    N = N_Werte[i]
    
    
    H = Hilbert(N)
    abweichungen[i] = maximum(abs.(norm(H * inv(H) - I(N))))
    
end
```

Plotten Sie nun wieder `abweichungen` gegen `N_Werte` .

```julia
plot(N_Werte, abweichungen)
```
