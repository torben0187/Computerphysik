Die **Zahl ${\pi}$** fasziniert die Menschen seit geraumer Zeit. Ihre anschaulichste Bedeutung ist wohl, dass sie das Verhältnis des Flächeninhalts eines Quadrats zu dem eines Kreises mit gleichem Durchmesser wie die Seitenlänge des Quadrats angibt. Analytisch berechnen lässt Sie sich auf verschiedenen Wegen im Prinzip beliebig genau.

Wir wollen in dieser Aufgabe eine Methode kennenlernen, mit der Sie $\pi$ auch dann berechnen könnten, wenn Sie einmal auf einer einsamen Insel stranden, alles über $\pi$ vergessen haben, aber unbedingt den Flächeninhalt eines Kreises berechnen müssen.

### a) Berechnung von $\pi$

Um in unserer Insel-Metapher zu bleiben: Stellen Sie sich vor, Sie zeichnen ein Quadrat in den Sand am Strand und darin einen Kreis. Sie werfen nun **Kieselsteine** in das Quadrat und schauen, wie oft Sie in den Kreis treffen (und wie oft nicht). Angenommen Sie könnten so werfen, dass Sie jeden Ort im Quadrat mit gleicher Wahrscheinlichkeit treffen, so ist es leicht nachzuvollziehen, dass das Verhältnis aus Kieseln im Kreis zu den insgesamt geworfenen Kieseln proportional zum Verhältnis ihrer Flächeninhalte und damit auch proportional zu $\pi$ ist. Berechnen Sie $\pi$ auf diesem Weg.

**Herleitung / Vorgehen:**

Wir betrachten ein Quadrat mit Seitenlänge $2r$, in das ein Kreis mit Radius $r$ einbeschrieben ist (Mittelpunkt im Ursprung). Es gilt:

$$A_{\text{Quadrat}} = (2r)^2 = 4r^2, \qquad A_{\text{Kreis}} = \pi r^2$$

$$\Rightarrow \frac{A_{\text{Kreis}}}{A_{\text{Quadrat}}} = \frac{\pi r^2}{4 r^2} = \frac{\pi}{4}$$

Wenn wir $N$ Punkte (Kieselsteine) gleichverteilt im Quadrat "werfen" und $N_{\text{Kreis}}$ davon innerhalb des Kreises landen, so gilt im Grenzfall großer $N$ (Gesetz der großen Zahlen):

$$\frac{N_{\text{Kreis}}}{N} \approx \frac{A_{\text{Kreis}}}{A_{\text{Quadrat}}} = \frac{\pi}{4} \quad\Rightarrow\quad \pi \approx 4\,\frac{N_{\text{Kreis}}}{N}$$

**Umsetzung (Monte-Carlo-Methode):**

1. Wir wählen $r=1$ (Einheitskreis) und ziehen $N$ zufällige Punkte $(x,y)$ gleichverteilt aus $[-1,1]\times[-1,1]$.
2. Ein Punkt liegt im Kreis, wenn $x^2 + y^2 \le 1$.
3. Wir zählen die Treffer $N_{\text{Kreis}}$ und werten $\pi \approx 4 N_{\text{Kreis}}/N$ aus.

```julia
using Random
using Statistics
using Plots

"""
    monte_carlo_pi(N)

Schätzt π, indem N zufällige Punkte gleichverteilt in [-1,1]x[-1,1]
gezogen und der Anteil der Punkte innerhalb des Einheitskreises bestimmt wird.
"""
function monte_carlo_pi(N::Integer)
    x = 2 .* rand(N) .- 1   # gleichverteilt in [-1, 1]
    y = 2 .* rand(N) .- 1   # gleichverteilt in [-1, 1]

    treffer = count(i -> x[i]^2 + y[i]^2 <= 1, 1:N)

    pi_schaetzung = 4 * treffer / N
    return pi_schaetzung
end

N = 1_000_000
pi_geschaetzt = monte_carlo_pi(N)

println("Anzahl Kieselsteine N    = ", N)
println("Geschätztes π            = ", pi_geschaetzt)
println("Tatsächliches π          = ", π)
println("Absoluter Fehler         = ", abs(pi_geschaetzt - π))
```

Zur Veranschaulichung plotten wir zusätzlich die geworfenen Kieselsteine: Punkte innerhalb des Kreises in einer Farbe, Punkte außerhalb in einer anderen.

```julia
N_plot = 2000
x = 2 .* rand(N_plot) .- 1
y = 2 .* rand(N_plot) .- 1
innen  = x.^2 .+ y.^2 .<= 1

scatter(x[innen],  y[innen],  ms=2, label="im Kreis",  color=:dodgerblue, aspect_ratio=:equal)
scatter!(x[.!innen], y[.!innen], ms=2, label="außerhalb", color=:orange)
θ = range(0, 2π, length=200)
plot!(cos.(θ), sin.(θ), color=:black, lw=2, label="Einheitskreis")
title!("Monte-Carlo-Bestimmung von π (N = $N_plot)")
```

### b) Fehler der Methode

Testen Sie wie sich der **Fehler** im Bezug auf die Anzahl an geworfenen Kieseln verhält. Plotten Sie dafür die mittlere absolute Abweichung Ihres Sampling von $\pi$ (z.B. bei 100 verschiedenen Samplings gleicher Kiesel-Anzahl) gegen die Anzahl der geworfenen Elemente in einem doppelt logarithmischen Plot und erklären Sie den Kurvenverlauf mit Ihrem Wissen aus der Vorlesung!

**Vorgehen:**

Für mehrere Werte von $N$ (z.B. logarithmisch verteilt von $N=10$ bis $N=10^6$) wiederholen wir die Monte-Carlo-Schätzung jeweils $M=100$-mal mit unabhängigen Zufallszahlen. Für jedes $N$ berechnen wir die **mittlere absolute Abweichung (MAD)**

$$\text{MAD}(N) = \frac{1}{M}\sum_{i=1}^{M} \left| \pi_i^{(N)} - \pi \right|$$

und tragen $\text{MAD}(N)$ gegen $N$ in einem doppelt-logarithmischen Plot auf.

```julia
# Logarithmisch verteilte Werte für N (z.B. 10^1 bis 10^6)
Ns = round.(Int, 10 .^ range(1, 6, length=20))
M  = 100   # Anzahl unabhängiger Samplings pro N

mad_values = zeros(length(Ns))

for (idx, N) in enumerate(Ns)
    fehler = [abs(monte_carlo_pi(N) - π) for _ in 1:M]
    mad_values[idx] = mean(fehler)
end

mad_values
```

```julia
plot(Ns, mad_values,
     xscale=:log10, yscale=:log10,
     marker=:circle, label="Simulation",
     xlabel="Anzahl Kieselsteine N",
     ylabel="mittlere absolute Abweichung |π_N - π|",
     title="Fehlerverhalten der Monte-Carlo-Schätzung")

# Vergleich mit dem theoretisch erwarteten Verhalten ~ 1/sqrt(N)
C = mad_values[1] * sqrt(Ns[1])     # Normierung an den ersten Punkt
plot!(Ns, C ./ sqrt.(Ns), ls=:dash, label="~ 1/√N")
```

**Erklärung des Kurvenverlaufs:**

Jeder geworfene Kieselstein kann als unabhängige Bernoulli-Zufallsvariable $X_i$ aufgefasst werden:

$$X_i = \begin{cases} 1, & \text{Stein } i \text{ landet im Kreis} \\ 0, & \text{sonst} \end{cases}, \qquad p = \Pr(X_i = 1) = \frac{\pi}{4}$$

Die Schätzung lautet $\hat\pi_N = 4\,\bar X_N$ mit dem Stichprobenmittel $\bar X_N = \frac{1}{N}\sum_{i=1}^N X_i$.

Nach dem **zentralen Grenzwertsatz** ist $\bar X_N$ für große $N$ näherungsweise normalverteilt mit

$$\text{Var}(\bar X_N) = \frac{\text{Var}(X_i)}{N} = \frac{p(1-p)}{N}$$

Die Standardabweichung (und damit auch die typische/mittlere absolute Abweichung) skaliert daher wie

$$\sigma(\bar X_N) \propto \frac{1}{\sqrt{N}} \quad\Rightarrow\quad \sigma(\hat\pi_N) = 4\,\sigma(\bar X_N) \propto \frac{1}{\sqrt{N}}$$

Das ist genau das allgemeine **Monte-Carlo-Fehlerverhalten** $\propto N^{-1/2}$, unabhängig von der Dimension des Problems – ein zentraler Vorteil von Monte-Carlo-Methoden gegenüber klassischen Integrationsverfahren in hohen Dimensionen.

Im doppelt-logarithmischen Plot äußert sich dieses Potenzgesetz als **Gerade mit Steigung $-1/2$**:

$$\log(\text{MAD}) = \log(C) - \frac{1}{2}\log(N)$$

Genau dieses Verhalten sehen wir in der Simulation: Die berechneten Datenpunkte folgen sehr gut der eingezeichneten $1/\sqrt{N}$-Referenzkurve. Um die Genauigkeit der Schätzung um einen Faktor 10 zu verbessern, muss man also die Anzahl der Stichproben (Kieselsteine) um den Faktor $10^2 = 100$ erhöhen – die Methode konvergiert vergleichsweise langsam, ist dafür aber sehr einfach und allgemein anwendbar (auch in hochdimensionalen Integrationsproblemen, wo deterministische Methoden ineffizient werden).
