<td style="border-style: none; width: 1%; text-align: left; font-size: 25px; font-weight: bold; text-decoration: underline">Aufgabe 1: Dendriten Wachstum</td>


Ver&auml;stelte Kristallstrukturen - sogenannte Dendriten - entstehen durch die **zuf&auml;llige Anlagerung diffusiver Teilchen**. Den Wachstumsprozess eines solchen Dendriten wollen wir in dieser Aufgabe illustrieren, indem wir den Zufallslauf einzelner diffusiver Teilchen simulieren. 

Dazu gehen wir wie folgt vor: Gegeben sei eine Experimentierscheibe von Radius $R$, in deren Mitte sich ein *Seed*,  ein erstes unbewegliches Teilchen, befinde. Wir starten nun ein weiteres Teilchen an beliebiger Position auf dem Rand der Experimentierscheibe. In jedem Simulationsschritt kann sich das Teilchen mit gleicher Wahrscheinlichkeit in eine von vier Richtungen (oben/unten/links/rechts) bewegen. Diese diffusive Bewegung simulieren wir so lange bis das Teilchen an den bestehenden Kristall andockt, d.h. es erreicht eine Position in deren unmittelbaren Nachbarschaft (oben/unten/links/rechts) sich ein bereits fixiertes, unbewegliches Teilchen befindet. In einem solchen Fall fixieren wir die Position des Teilchens und starten das n&auml;chste Teilchen vom Rand. Sollte das Teilchen w&auml;hrend der Simulation von der Experimentierscheibe wandern -- also einen Abstand gr&ouml;&szlig;er als R vom Mittelpunkt erreichen  dann verwerfen wir das aktuelle Teilchen und starten ein neues Teilchen auf dem Rand.

Simulieren Sie den Wachstum eines derartigen Dendriten auf einem Raster von $128 \times 128$ Positionen und einer Experimentierscheibe vom Radius $R=64$. Stellen Sie den gewachsene Struktur graphisch dar.

Wer sich ein wenig &uuml;ber die historische Entwicklung dieses im Englischen als *diffusion limited aggregation* bezeichneten Ph&auml;nomens informieren m&ouml;chte, der sei auf diesen [Bericht](http://garfield.library.upenn.edu/classics1993/A1993KX31200001.pdf) seines Entdeckers Thomas Witten von der University of Chicago verwiesen.

### L&ouml;sungsidee

Wir legen ein quadratisches Gitter `schale` der Gr&ouml;&szlig;e $128\times128$ an, das mit Nullen initialisiert wird. Eine `1` an Position `(i,j)` bedeutet: dort sitzt ein fixiertes (angedocktes) Teilchen. Der Mittelpunkt des Gitters bekommt den Seed.

Die Funktion `add_particle!` macht Folgendes:

1. **Start auf dem Rand:** Eine zuf&auml;llige Position auf dem Kreis vom Radius $R$ um den Mittelpunkt wird bestimmt (das ist im Geger&uuml;st schon vorgegeben).
2. **Random Walk:** In einer Schleife w&uuml;rfeln wir bei jedem Schritt eine von vier Richtungen (oben/unten/links/rechts) mit gleicher Wahrscheinlichkeit und bewegen das Teilchen entsprechend.
3. **Abbruchbedingung "Andocken":** Nach jedem Schritt pr&uuml;fen wir, ob einer der vier Nachbarn (oben/unten/links/rechts) der neuen Position bereits besetzt ist. Falls ja, wird die aktuelle Position in `schale` auf `1` gesetzt und die Schleife (`transport = false`) beendet.
4. **Abbruchbedingung "Scheibe verlassen":** Wir berechnen nach jedem Schritt den Abstand zum Mittelpunkt. Ist dieser gr&ouml;&szlig;er als $R$, verwerfen wir das Teilchen und starten es neu auf dem Rand (anstatt die ganze Funktion neu aufzurufen, was die a&uuml;&szlig;ere Wachstumsschleife unn&ouml;tig oft aufrufen w&uuml;rde).

Anschlie&szlig;end rufen wir `add_particle!` sehr oft in einer Schleife auf ("Wachstum") und stellen das Endergebnis mit `imshow` dar.

```julia
# funktion zum teilchen anlagern
function add_particle!(schale)
    # radius bestimmen
    R = size(schale)[1]/2 - 1
    
    # zufällige position auf dem rand bestimmen
    theta = rand()*2*pi
    i = round(Int64, floor(cos(theta)*R + R+1))
    j = round(Int64, floor(sin(theta)*R + R+1))
    
    # bewege das teilchen solange es nicht fest haftet
    transport = true
    
    
    ###########################
    ##hier kommt ihr code hin##
    ###########################
    
    # Mittelpunkt des Gitters (entspricht dem Seed)
    center = R + 1
    
    while transport
        # zufällige Richtung würfeln: 1=oben, 2=unten, 3=rechts, 4=links
        richtung = rand(1:4)
        if richtung == 1
            i += 1
        elseif richtung == 2
            i -= 1
        elseif richtung == 3
            j += 1
        else
            j -= 1
        end
        
        # Abstand zum Mittelpunkt prüfen -> Teilchen die Scheibe verlassen?
        abstand = sqrt((i - center)^2 + (j - center)^2)
        if abstand > R
            # Teilchen verwerfen und neu auf dem Rand starten
            theta = rand()*2*pi
            i = round(Int64, floor(cos(theta)*R + R+1))
            j = round(Int64, floor(sin(theta)*R + R+1))
            continue
        end
        
        # prüfen, ob ein Nachbar (oben/unten/links/rechts) schon besetzt ist
        if schale[i+1, j] == 1 || schale[i-1, j] == 1 ||
           schale[i, j+1] == 1 || schale[i, j-1] == 1
            # Teilchen andocken: Position fixieren
            schale[i, j] = 1
            transport = false
        end
    end
end
```

Nun lassen wir den Dendriten wachsen, indem wir `add_particle!` viele Male aufrufen, und stellen das Ergebnis grafisch dar.

```julia
###########################################
##hier Wachstum mit plot veranschaulichen##
###########################################
using PyPlot

# Gittergröße und Seed initialisieren
N = 128
schale = zeros(Int64, N, N)

center = round(Int64, N/2)
schale[center, center] = 1   # Seed in der Mitte

# Anzahl der Teilchen, die angelagert werden sollen
n_particles = 1500

for k in 1:n_particles
    add_particle!(schale)
end

# Darstellung der gewachsenen Struktur
figure(figsize=(6,6))
imshow(schale, cmap="inferno", origin="lower")
title("Dendritenwachstum (Diffusion Limited Aggregation), N=$n_particles Teilchen")
axis("off")
savefig("dendrit.png", dpi=150)
gcf()
```

### Anmerkungen

- Mit `R = size(schale)[1]/2 - 1` und Gittergr&ouml;&szlig;e $128$ ergibt sich $R = 63$, also fast genau der geforderte Radius $R=64$ &ndash; das passt zur Vorgabe im Aufgabengeger&uuml;st.
- Die Anzahl `n_particles = 1500` ist ein guter Kompromiss: Genug Teilchen, um eine deutlich verzweigte (dendritische) Struktur zu erkennen, aber noch in vertretbarer Rechenzeit (der Random Walk eines Teilchens kann je nach Position lange dauern, da er erst beim Andocken stoppt). Wer m&ouml;chte, kann den Wert erh&ouml;hen, bis die Scheibe sichtbar dichter zuw&auml;chst &ndash; die Simulation wird dann allerdings deutlich langsamer, da sp&auml;te Teilchen oft sehr lange Zufallswege zur&uuml;cklegen, bevor sie andocken.
- Charakteristisch f&uuml;r *diffusion limited aggregation* ist die fraktale, fingerartig verzweigte Struktur: Spitzen, die zuerst von ankommenden Teilchen erreicht werden, "fangen" diese bevorzugt ein, wodurch sie schneller wachsen als die L&uuml;cken dazwischen (Abschirmungseffekt). Genau dieses Verhalten sollte man im erzeugten Bild erkennen.
