### c) `for` - Schleifen

Letzte Woche haben Sie in den Programmiertutorials schon die `while`-Schleife kennengelernt. Diese Woche wird dort auch die meist noch häufiger benutzte `for`-Schleife eingeführt. Diese kann verwendet werden, wenn Anweisungen für eine bekannte Anzahl an Wiederholungen ausgeführt werden sollen.

Schreiben Sie in der nachfolgenden Zelle eine `for`-Schleife, die die Quadrate der Zahlen 1 bis 10 ausgibt. 

```julia
zahlen=collect(1:10)
for i in zahlen
    print(i^2)
    if i<10
        print(",")
    end
end
```

Schreiben Sie nun eine `for`-Schleife, die wie oben die Quadrate der Zahlen 1 bis 10 ausgibt, doch diesmal nur für gerade Zahlen.

```julia
zahlen=collect(1:10)
for i in zahlen
    if i%2==0
        print(i^2)
        if i<10
            print(",")
        end
    end
end
```

Ergänzen Sie anschließend den obigen Code, sodass die Summe der Quadrate in einer Variable `summe` gespeichert und nach der Schleife ausgegeben wird.

```julia
zahlen=collect(1:10)
summe=0
for i in zahlen
    if i%2==0
        print(i^2)
        summe=summe+i^2
        if i<10
            print(",")
        end
    end
end
print(" | Summe: ",summe)
```

### d) `while` - Schleifen

`while`-Schleifen laufen im Gegensatz zur `for`-Schleifen normalerweise nicht für eine im vorhinein bekannte Anzahl an Wiederholungen, sondern so lange, bis eine `<Bedingung>` erfüllt ist.  
Ergänzen Sie die nachfolgende Zelle um eine `while`-Schleife, die die Variable `M` in jedem Schritt um 10% veringert und anschließend den neuen Wert ausgibt. Die Schleife soll so lange laufen, bis M kleiner gleich 5 ist.  

```julia
M = 10
print(M,", ")
while (M)>=5
    M=M*0.9
    print(M,", ")
end
```

Nun soll eine Funktion `quadrate(N)` geschrieben werden, die einen Array mit den Quadraten der ganzen Zahlen von 1 bis N ausgibt. Ergänzen Sie dafür die untere Funktionsdefinition, sodass so ein Array erstellt, befüllt, und anschließend mit `return` ausgegeben wird. Testen Sie anschließend ihre Funktion mit verschiedenen Werten von `N`.

```julia
function quadrate(N)
    array=zeros(N)
    for i in 1:N
    array[i]=i^2
    end
return array
end

    
```

```julia
quadrate(5)
```
