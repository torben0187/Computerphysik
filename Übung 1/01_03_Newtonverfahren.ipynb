Ein weiteres häufig verwendetes Verfahren zur Nullstellenbestimmung ist das Newtonverfahren, auch Newton-Raphson-Verfahren genannt. Anders als das Bisektionsverfahren nutzt das Newtonverfahren nicht nur die Funktion selbst, sondern auch die Ableitung der Funktion aus. Im folgenden sollen die das Newtonverfahren implementieren und anschließend zur Nullstellensuche verwenden.

### a) Grundidee

Das Newtonverfahren berechnet für einen gegebenen Startwert $x_0$ und eine Funktion $f$ die Folge 

$$x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}.$$

Diese Folge konvergiert für geignete Funktionen gegen deren Nullstelle. Schildern Sie im Folgenden kurz die Grundidee des Newtonverfahrens und beschreiben Sie, wie daraus obige Formel hergeleitet werden kann.

### b) Implementierung 

Für die Implementierung des Newtonverfahrens muss auch die Ableitung einer Funktion berechnet werden. Wie dies numerisch effizient und präzise gemacht werden kann, wird im Vorlesungsteil zur numerischen Differentiation noch genauer beschrieben. Hier genügt es, wenn Sie die folgende Funktion `diff_quotient(f, x, epsilon)` verwenden. Sie  berechnet die Ableitung einer Funktion `f` an der Stelle `x` mit dem maximalen Fehler `epsilon` mit Hilfe des Differenzenquotienten. Führen sie zunächst die nachfolgende Zelle aus.

```julia
function diff_quotient(f, x, epsilon = 1e-8)
    return (f(x + epsilon) - f(x))/epsilon
end

#Test:
f(x) = x^2 + 1
println("Die Ableitung von f an der Stelle x=3 lautet ungefähr: ", diff_quotient(f, 3))
```
