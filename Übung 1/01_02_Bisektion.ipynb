### a) Implementieren der Bisektion

Die Vorgehensweise bei der Bisektion ist folgende:  

Es soll die Nullstelle einer Funktion $f$ im Intervall $[x_l, x_r]$ gefunden werden. Dafür berechnet man zunächst die Mitte des Intervals $x_m$. Dann überprüft man, ob sich das Vorzeichen der Funktion von der linken Intervallgrenze $x_l$ zur Mitte $x_m$, oder von der Mitte $x_m$ zur rechten Intervallgrenze $x_r$ ändert. Dort, wo sich das Vorzeichen ändert, muss zwingend eine Nullstelle vorhanden sein. Anschließend wird die Prozedur in diesem Intervall erneut wiederholt. Dies geschieht so lange, bis eine gewünschte Präzision erreicht ist. 

Vervollständigen Sie die folgende Funktion um die Nullstellensuche mittels der Bisektion. Die Argumente der Funktion sind

- `f` die (mathematische) Funktion, deren Nullstelle bestimmt werden soll
- `x_left` und `x_right` das Intervall in dem die Nullstelle gesucht wird
- `precision` Genauigkeit der Nullstellenbestimmung, d.h. inwiefern der Wert `f(x_0)` von `0` abweichen darf
- `max_iterations` maximale Anzahl an Bisektions-Schritten

Der Aufbau der Funktion ist dabei schon weitgehend vorgegeben. Sie müssen noch die Berechnung der Intervallmitte und das Überprüfen auf Vorzeichenwechsel und dementsprechende Anpassen der Intervallgrenzen implementieren.

```julia
function bisektion(f, x_left, x_right, precision, N=100)
    x_mitte = 0
    for i in 1:N
        x_mitte = (x_left+x_right)/2 
        f_mitte = f(x_mitte)
        if abs(f_mitte) < precision
            return x_mitte
        end
        if f(x_left)*f_mitte>0
            x_left=x_mitte
        else
            x_right=x_mitte
        end       
    end
    return x_mitte
end
```

Testen Sie Ihren Code, indem Sie versuchen eine Nullstelle der Funktion $f(x) = 2x^2 - 8$ zwischen 0 und 5 bei einer Präzision von 1e-3 zu finden. Kann Ihr Code die Nullstelle richtig identifizieren?
