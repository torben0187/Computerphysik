### Aufgabe 1: Numerische Stabilitaet von Zeitintegration

---

In dieser Aufgabe soll die numerische Stabilitaet von verschiedenen Zeitintegrationsschemata untersucht werden:
- Explizites Eulerverfahren
- Implizites Eulerverfahren
- Crank-Nicolson-Verfahren

Diese 3 Verfahren sollen anhand der Diffusionsgleichung

$$
\partial_t u(x,t) = D\partial_x^2 u(x,t)
$$

$$
u(x_{min},t) = u(x_{max},t) = 0
$$

mit 

$$
u: [x_{min},x_{max}]\times [0,T] \rightarrow \mathbb{R}
$$


numerisch untersucht werden.
Dazu wird das Problem wieder in ein Raum und Zeitgitter aufgeteilt, wobei wir
fuer den Raum 

$$
    x_j := x_{min} + \delta x (j-1)
$$

mit

$$
   \delta x := \frac{x_{max} - x_{min}}{N_x - 1}
$$

und fuer die Zeit

$$
    t_j :=\delta t (j-1)
$$

mit

$$
   \delta t := \frac{T}{N_t - 1}
$$

definieren. Dabei bezeichnet $T$ die maximal zu simulierende Zeit. Die numerische Approximation der exakten Loesung am Ort $t_n$ zum Zeitpunkt $x_j$ bezeichnen wir wieder mit $u^n_j$.
Mithilfe dieser Diskretisierung laesst sich die Diffusionsgleichung als Matrixgleichung schreiben, die den Vektor 
$\vec{u}^n := \left( u_1^n,\dots, u_{N_x}^n\right)$ aller Ortskoordinaten zu dem Zeitpunkt $t_n$, mit dem Vektor bei $t_{n+1} = t_n + \delta t$ verknuepft:
$$
    (1 - a\delta t  L) \vec{u}^{n+1} =  (1 + b\delta t  L) \vec{u}^{n} \,.
$$

Dabei bezeichnet $L$ die Matrix, die die diskretisierte Form von $D\partial_x^2$ darstellt:

$$
\left(L\vec{u}^n\right)_j = D \frac{u^n_{j+1} - 2u^n_{j} + u^n_{j-1}}{\delta x^2} \, .
$$

mit

$$
\left(L\vec{u}^n\right)_0 = \left(L\vec{u}^n\right)_{N_x} = 0
$$

um die Dirichlet Randbedingungen zu erfuellen.

Die konkrete Wahl der Parameter $a$ und $b$ liefern dann unterschiedliche Verfahren, wobei hier die folgenden zu betrachten sind:
- Explizit Euler: $a = 0,\ b=1$
- Implizit Euler: $a = 1,\ b=0$
- Crank-Nicolson: $a=1/2,\ b = 1/2$

Desweiteren werden wir im folgenden $D = 1$ annehmen.

---

### a) Generation der Matrix  L

---

Erstellen Sie eine Funktion, die die Diskretisierungsmatrix des Diffusionsoperators $\partial_x^2$ zurueckgibt. Versuchen Sie die Matrix im CSC Format zu implementieren!

```julia
using SparseArrays

function generate_L(Nx, dx)
    main_diag = -2 .* ones(Nx)
    off_diag = ones(Nx-1)

    L = spdiagm(
        -1 => off_diag,
         0 => main_diag,
         1 => off_diag 
    ) * D / dx^2
    return L
end
```

### b) Implementierung der Zeitintegrationsverfahren

---
Implementieren Sie nun 3 Funktionen, die nach den oben genannten Methoden einen Zeitschritt ausfuehren. Beachten Sie, dass man im Falle des impliziten Euler und Crank-Nicolson Verfahren ein Lineares Gleichungssystem der Form

$$
B\vec{u}^{n+1} = A\vec{u}^n 
$$

loesen muss. Die Funktionshuelsen sind schon vorgegeben, da die entsprechenden Funktionen spaeter verwendet werden um eine Simulation mithilfe einer Animation zu visualisieren. Jede Methode bekommt die aktuelle Konfiguration
`u_n` uebergeben, welches ein Array der Laenge $N_x$ ist, zusammen mit der Orts- und Zeitdiskretisierung `dx` bzw `dt`. Zurueckgegeben werden soll die Konfiguration zum naechsten Zeitpunkt `u_next`.

```julia
using LinearAlgebra

function euler_exp(u_n,dx,dt)
    N_x = length(u_n)
    u_next = ones(N_x)
    u_next = u_n + dt * (L * u_n)
    return u_next
end

function euler_imp(u_n,dx,dt)

    N_x = length(u_n)
    B = Matrix(I, N_x, N_x) - dt*L
    u_next = B \ u_n
    return u_next
end;

function crank_nicolson(u_n,dx,dt)
    N_x = length(u_n)
    A = Matrix(I, N_x, N_x) + 1/2 * dt * L
    B = Matrix(I, N_x, N_x) - 1/2 * dt * L
    u_next = B \ (A * u_n)
    return u_next
end;
```

#### c) Stabilisationstest

---

Wenn man sich die numerische Stabilitaet der Algorithmen anschaut, so faellt auf, dass das explizite Eulerverfahren nur fuer 

$$
    \frac{\delta t}{\delta x^2} < \frac{1}{2}
$$

ein stabiles Ergebnis liefern kann. Dies soll im folgenden untersucht werden. Dafuer nehmen wir ein System mit
- x_min = 0
- x_max = 1
- u(x,0) = sin(pi*x)

an, fuer welches die exakte Loesung der Diffusionsgleichung als

$$
    u_{ana}(x,t) = e^{-\pi^2 t} \sin(\pi x)
$$

bekannt ist. Nutzen Sie die untenstehende `calculate_time_evolution` Methode, welche eine Anfangskonfiguration `u_0`, das Raumgitter `x_range` und `t_range` uebergeben bekommt und dann mit der gewaehlten Methode die gesamte Zeitevolution berechnet. Die Methode wird ueber den optionalen Parameter `method` uebergeben und kann
- euler_exp
- euler_imp
- crank_nicolson

sein. Je nach gewaehlter Methode wird die von Ihnen geschriebene Methode fuer einen Zeitschritt aufgerufen, um die Konfiguration im naechsten Zeitschritt zu berechnen. Das Ergebnis wird dann in einer $N_x \times N_t$ matrix zurueckgegeben.

```julia
function calculate_time_evolution(u_0, x_range, t_range; method= "euler_exp")
    Nx = size(x_range,1)
    Nt = size(t_range,1)
    δt = t_range[2] - t_range[1]
    δx = x_range[2] - x_range[1]
    
    # zeitentwicklung als matrix erstellen
    time_evolution = zeros(Nx, Nt)
    # setze die anfangsbedingungen, und speicher die erste zeitentwicklung in die matrix
    time_evolution[:,1] .= u_0
    
    for tt in 2:Nt
        if(method == "euler_exp")
            time_evolution[:,tt] .= euler_exp(time_evolution[:,tt-1], δx,δt)
        elseif(method == "euler_imp")
            time_evolution[:,tt] .= euler_imp(time_evolution[:,tt-1], δx,δt)
        elseif(method == "crank_nicolson")
            time_evolution[:,tt] .= crank_nicolson(time_evolution[:,tt-1], δx,δt)
        end
    end
    return time_evolution
end;
```
