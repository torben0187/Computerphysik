<td style="border-style: none; width:1%; text-align: left; font-size: 25px; font-weight: 200">Relaxen im Plattenkondensator</td>


Den Plattenkondensator und sein elektrisches Feld haben Sie schon in den einführenden Vorlesungen zur Experimentalphysik kennengelernt. Meist wird dabei ein homogenes Feld innerhalb des Kondensators angenommen, mit Feldlinien, die wie unten gezeigt verlaufen. 

In dieser Aufgabe wollen wir überprüfen, in wie weit diese Annahme zutrifft. Dazu wollen wir das Feld des Plattenkondensators aus der Lösung der **Laplace-Gleichung**

\begin{equation}
\Delta\phi = 0
\end{equation}

berechnen, indem wir auf die in der Vorlesung vorgestellte **Relaxationsmethode** zurückgreifen. Die Randbedingungen seien dabei so gewählt, dass die beiden Kondensatorplatten auf einem Potential von $+1$ bzw. $-1$ und der Rand der betrachteten Fläche auf einem Potential von $0$ liegt.
