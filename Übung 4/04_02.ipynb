In diesem Blatt verwenden wir den $QR$-Algorithmus, um die Eigenwerte einer Matrix $A$ zu ermitteln. Der Algorithmus verwendet iterativ die $QR$-Zerlegung, welche eine unitäre Matrix $Q$ und eine oberdreieckige $R$ liefert, um die Eigenbasis anzunähren, d.h., um die Matrix $A$ bei jedem Schritt *diagonaler* zu machen. Abgesehen von der theoretischen Hintergrunden, wollen wir hier dieses Verfahren implementieren und seine Leistungen einschätzen.

Angenommen sei eine $n \times n$ quadratische, hermitische Matrix $A$ i.e., $A^\dagger = A$. 

Legen Sie $A_0=A$, $V_0=\mathbb{I}_n$ fest, und gehen Sie iterativ wie folgt vor:

i) Führen Sie eine QR-Zerlegung von $A_{k}$ durch: $A_{k} = Q_{k+1} R_{k+1}$


ii) Aktualisieren Sie den Wert von $V$ als $V_{k+1} = V_k\,Q_{k+1}$


iii) Definieren Sie $A_{k+1} = R_{k+1} Q_{k+1} \left(=V_{k+1}^{\dagger}A_0V_{k+1}\right)$, was eine Ähnlichkeitstransformation darstellt. Das Eigenspektrum ist gleich für $A$ und alle $A_k$!

iv) Testen Sie, wie nah das Neue $A_k$ auf die gewünschte diagonale Form ist, d.h., bewerten Sie 

$$
\epsilon_k = \frac{\displaystyle \max_{i\neq j} \lvert A_{ij} \rvert}{\displaystyle \max_{j} \lvert A_{jj} \rvert}
$$

und vergleichen Sie es mit einem zuvor eingestellten Toleranzwert $\bar{\epsilon}$: 
Falls $\epsilon_k < \bar{\epsilon}$, dann verlassen Sie die iterative Prozedur und springen Sie zum letzten Schritt.

Wiederholen Sie den Vorgang jedenfalls höchstens bis zu einer zuvor festgelegten maximalen Anzahl von Schritten $\bar{k}$.


Sobald der iterative Prozess gestoppt ist, definieren Sie $k_{\max}$ als den zuletzt ausgewerteten Index und $\epsilon_{\max} = \epsilon_{k_{\max}}$.

Die so erhaltene Matrix ist ungefähr diagonal und Sie können die außerdiagonalen Elemente, die vernachlässigbar sind, auf folgende Weise rausfiltern:
$$\Lambda_{ij} = \delta_{ij} \,(A_{k_{\max}})_{jj}$$

Nun könnten Sie überprüfen, dass $Q^\dagger A Q = \Lambda$ bis zu einem Fehler vergleichbar mit $\epsilon_{\max}$ gilt, und evtl. auch wie nah die diagonale Elemente von $\Lambda$ zu den wahren Eigenwerten liegen.
