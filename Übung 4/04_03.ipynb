# Schwingenden Feder-Masse-Kette

Wir betrachten eine eindimensionale Kette aus $N$ gleichen Massen, die jeweils durch Federn mit ihren direkten Nachbarn verbunden sind.

Für kleine Auslenkungen aus der Ruhelage schreiben wir für die Kraft die auf die $i$-te Masse wirkt:

$$F_i = m \ddot{x}_i = k(x_{i+1} - 2x_i + x_{i-1})$$

Dabei ist

- $m$ die Masse eines Teilchens, im Folgeden $m = 1$,
- $k$ die Federkonstante,
- $x_i$ die Auslenkung der $i$-ten Masse,
- $x_{i-1}$ und $x_{i+1}$ die Auslenkungen der direkten Nachbarn.

Die Bewegungsgleichung zeigt: Jede Masse ist nur über die Federn mit ihren direkten Nachbarn gekoppelt. Wir können nun die Auslenkungen der verschiedenen Massen in einen Vektor schreiben $\vec{x}$ dessen $i$-ter genau $x_i$ enthält.

So lässt sich die obige Gleichung schreiben als 

$$
    m\ddot{\vec{x}} = -K \vec{x}
$$

Wobei $K$ ie Kopplungsmatrix ist und die folgende Form hat:

$$
K =
k
\begin{pmatrix}
2 & -1 & 0 & \dots & 0 \\
-1 & 2 & -1 & \dots & 0 \\
0 & -1 & 2 & \dots & 0 \\
\vdots & \vdots & \vdots & \ddots & -1 \\
0 & 0 & 0 & -1 & 2
\end{pmatrix}.
$$

Die Hauptdiagonale enthält den Wert $2k$, die beiden Nebendiagonalen enthalten den Wert $-k$. Alle anderen Einträge sind null.

Da für große $N$ fast alle Matrixeinträge null sind, ist diese Matrix ein gutes Beispiel für eine **sparse matrix**.

## Aufgabe 3: Eigenfrequenzen der Feder-Masse-Kette

Die Bewegungsgleichung der gekoppelten Feder-Masse-Kette lautet

$$
m \ddot{\vec{x}} = -K \vec{x}
$$.

Um die Schwingungen der Kette zu untersuchen, sucht man Lösungen der Form

$$
\vec{x}(t) = \vec{u} \cos(\omega t)
$$.

Hierbei ist

- $\vec{u}$ ein Eigenvektor der Bewegung,
- $\omega$ die Kreisfrequenz der Schwingung.

Setzt man diesen Ansatz in die Bewegungsgleichung ein, erhält man

$$
K \vec{u} = m \omega^2 \vec{u}
$$.

Dies ist ein Eigenwertproblem. Die Eigenwerte $\lambda$ der Matrix $K$ hängen mit den Eigenfrequenzen zusammen über

$$
\lambda = m \omega^2
$$.

Daraus folgt:

$$
\omega = \sqrt{\frac{\lambda}{m}}
$$.

Die Eigenfrequenzen beschreiben die **Normalschwingungen** der Kette. 

Verwenden Sie die sparse Kopplungsmatrix `K_sparse` aus den vorherigen Aufgaben.

Berechnen Sie einige der kleinsten Eigenwerte von `K_sparse` mit einem sparse Eigenwertsolver und bestimmen Sie daraus die zugehörigen Eigenfrequenzen.

## Aufgabe 4: (BONUSINFO) Matrixfreie Darstellung der Kopplungsmatrix

Bisher wurde die Kopplungsmatrix $K$ explizit als sparse matrix gespeichert. Für sehr große Systeme kann es aber noch effizienter sein, die Matrix gar nicht explizit zu erstellen.

Stattdessen definiert man nur eine Funktion, die die Wirkung der Matrix auf einen Vektor berechnet.

$K\vec{x} =
k
\begin{pmatrix}
2x_1 - x_2 \\
-x_1 + 2x_2 - x_3 \\
-x_2 + 2x_3 - x_4 \\
\vdots \\
-x_{N-1} + 2x_N
\end{pmatrix}$.

Das bedeutet: Um $K\vec{x}$ zu berechnen, muss man die Matrix $K$ nicht vollständig speichern. Man muss nur wissen, wie jeder Eintrag des Ergebnisvektors berechnet wird. Die folgende Funktion tut eben dies:

```julia
function K_action(x)
    y = similar(x)

    y[1] = k * (2x[1] - x[2])

    for i in 2:length(x)-1
        y[i] = k * (-x[i-1] + 2x[i] - x[i+1])
    end

    y[end] = k * (-x[end-1] + 2x[end])

    return y
end
```

```julia
x = rand(N)
y = K_action(x)

ytest = K_sparse * x

isapprox(y, ytest)
```

Mit dieser Action-Funktion lässt sich sogar das Lanzos- oder Arnoldi-Verfahren durchführen!

```julia
eigenvalues, eigenvectors, info = eigsolve(K_action, rand(N), 6, :SR, krylovdim = 60)
```
