In dieser Aufgabe beschäftigen wir uns mit dem Thema **Numerische Integration**. Aus der Analysis sind Ihnen **Riemann (Ober-)Summen** bekannt. Die erste neue Methode, die Sie kennengelernt haben, ist die **Trapezregel**. Wie der Name schon suggeriert, wird der Flächeninhalt durch eine Reihe von Trapezen approximiert. Abhängig vom Verhalten der Funktion, dem Diskretisierungsschritt und der Größe des Integrationsbereichs kann dies bereits gute Ergebnisse liefern. Genauer ist allerdings die **Simpson Regel**, bei der die Funktion auch diskretisiert, aber dann intervallweise durch eine Parabel genähert wird.

$$ 
\begin{align}
Riemann\ Summe &: \int_a^b f(x) \operatorname{dx} \approx \sum_{j=2}^N f(x_j)\Delta x \\
Trapez\ Regel &: \int_a^b f(x) \operatorname{dx} \approx \frac{1}{2}\sum_{j=2}^N \left( f(x_j) + f(x_{j-1}) \right) \Delta x \\
Simpson\ Regel &: \int_a^b f(x) \operatorname{dx} \approx \frac{1}{6}\sum_{j=2}^N \left( f(x_j) + 4f\left(\frac{x_j+x_{j-1}}{2}\right) + f(x_{j-1}) \right)\Delta x 
\end{align}
$$
