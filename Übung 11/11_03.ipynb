```julia
using Plots

function animate_walker(dist = () -> rand(1:5); N=20)
    pos, Mat = init(N)
    
    anim = @animate for t in 1:200
        walker_step!(pos, Mat, dist)
        heatmap(Mat, c = :grays)
    end
    
    return gif(anim, "walker.gif", fps = 30)    
end
```
